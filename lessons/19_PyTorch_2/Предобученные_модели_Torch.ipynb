```python
from torchvision import transforms
preprocess = transforms.Compose([
transforms.Resize(256),
transforms.CenterCrop(224),
transforms.ToTensor(),
transforms.Normalize(
mean=[0.485, 0.456, 0.406],
std=[0.229, 0.224, 0.225]
)])
```

```python
!wget -r https://github.com/ankislyakov/data/blob/main/bpla.jpg?raw=true
```

```python
from PIL import Image
img = Image.open("/content/github.com/ankislyakov/data/blob/main/bpla.jpg?raw=true")
```

```python
img
```

```python
img_t = preprocess(img)
```

```python
img_t
```

```python
batch_t = torch.unsqueeze(img_t, 0)
```

```python
resnet.eval()
```

```python
out = resnet(batch_t)
out
```

```python
!wget -r https://raw.githubusercontent.com/pytorch/hub/master/imagenet_classes.txt
```

```python
with open('/content/raw.githubusercontent.com/pytorch/hub/master/imagenet_classes.txt') as f:
    labels = [line.strip() for line in f.readlines()]
```

```python
_, index = torch.max(out, 1)
```

```python
percentage = torch.nn.functional.softmax(out, dim=1)[0] * 100
labels[index[0]], percentage[index[0]].item()
```

```python
_, indices = torch.sort(out, descending=True)
[(labels[idx], percentage[idx].item()) for idx in indices[0][:5]]
```
